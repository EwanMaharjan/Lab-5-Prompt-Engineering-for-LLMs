```python
import os, random, textwrap
import torch
from transformers import pipeline, AutoModelForCausalLM, AutoTokenizer

# Reproducibility
SEED = 42
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device selection
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)
```

```python
# choosing a model strategy, picked on of 3 from the lab
```

```python
INSTRUCT_MODEL_ID = "microsoft/Phi-3-mini-4k-instruct"

instruct_tokenizer = AutoTokenizer.from_pretrained(INSTRUCT_MODEL_ID)
instruct_model = AutoModelForCausalLM.from_pretrained(
    INSTRUCT_MODEL_ID,
    torch_dtype="auto",
    device_map="auto" if DEVICE == "cuda" else None
)
instruct_generator = pipeline(
    task="text-generation",
    model=instruct_model,
    tokenizer=instruct_tokenizer,
    return_full_text=False,
)
print("Loaded:", INSTRUCT_MODEL_ID)
```

```python
#helper function (pretty print + safe generation)
```

```python
def show(title, text):
    """Nicely formatted section header + wrapped text."""
    print("\n" + "="*90)
    print(title)
    print("-"*90)
    print(text)
    print("="*90)

def safe_generate(gen_pipeline, prompt, max_new_tokens=80, temperature=0.7, top_p=0.95,
do_sample=True):
    """Wrapper around a text-generation pipeline to avoid common pitfalls."""
    try:
        tokenizer = gen_pipeline.tokenizer
        pad_token_id = tokenizer.pad_token_id or tokenizer.eos_token_id
    except Exception:
        pad_token_id = None

    kwargs = dict(
    max_new_tokens=max_new_tokens,
    temperature=temperature,
    top_p=top_p,
    do_sample=do_sample,
    return_full_text=True
    )
    if pad_token_id is not None:
        kwargs["pad_token_id"] = pad_token_id

    output = gen_pipeline(prompt, **kwargs)[0]["generated_text"]
    return output

def try_prompt(label, gen_pipeline, prompt, **gen_kwargs):
    """Run a prompt through a chosen pipeline and pretty-print the result."""
    out = safe_generate(gen_pipeline, prompt, **gen_kwargs)
    show(f"{label} PROMPT", prompt)
    show(f"{label} OUTPUT", out)
```

```python
# EXERCISE 1 - ZERO SHOT, ONE SHOT AND FEW SHOT PROMPTING
```

```python
# Helper function to run prompts with clear labels
def try_prompt(label, prompt):
    out = safe_generate(GEN, prompt, max_new_tokens=80, temperature=0.8)
    show(f"{label} PROMPT", prompt)
    show(f"{label} OUTPUT", out)

# -------------------------------
# Zero-Shot Prompt
# -------------------------------
p_zero = (
"Classify the sentiment of this review in one word (Positive, Negative, or Neutral): "
"'The product broke after one day.'"
)
try_prompt("ZERO-SHOT", p_zero)

# -------------------------------
# One-Shot Prompt
# -------------------------------
p_one = (
"You are a sentiment classifier.\n"
"Example: 'I love this!' Positive\n"
"Now classify in ONE WORD (Positive, Negative, or Neutral): "
"'The product broke after one day.'"
)
try_prompt("ONE-SHOT", p_one)

# -------------------------------
# Few-Shot Prompt
# -------------------------------
p_few = (
"You are a sentiment classifier.\n"
"Examples:\n"
" - 'I love this!' Positive\n"
" - 'This is terrible.' Negative\n"
" - 'It works as expected.' Neutral\n"
"Now classify in ONE WORD (Positive, Negative, or Neutral): "
"'The product broke after one day.'"
)

try_prompt("FEW-SHOT", p_few)
```

```python
results = {}
for label, prompt in {
    "Zero-Shot": p_zero,
    "One-Shot": p_one,
    "Few-Shot": p_few
}.items():
    results[label] = safe_generate(GEN, prompt, max_new_tokens=80, temperature=0.7)

for k, v in results.items():
    show(f"{k} OUTPUT", v)
```

```python
p_few_structured = (
"Instruction: Classify the following review as Positive, Negative, or Neutral.\n"
"Context: You are analyzing customer satisfaction for a product.\n"
"Constraints: Respond in ONE word only.\n"
"Examples:\n"
" - 'Amazing service!' Positive\n"
" - 'Worst purchase ever.' Negative\n"
"Output Format: Single word only.\n"
"Review: 'The product broke after one day.'"
)
try_prompt("FEW-SHOT + STRUCTURED", p_few_structured)
```

```python
#EXERCISE 2 - STRUCTURED PROMPT DESIGN
```

```python
# Paragraph for summarization
paragraph = (
"Artificial intelligence (AI) systems are increasingly used in education, finance, and healthcare. "
"However, when algorithms are poorly designed or trained on biased data, they may amplify inequality "
"and make unfair decisions. Building responsible AI requires fairness audits, transparency, and human oversight."
)

# 1 Unstructured prompt
p_unstructured = f"Summarize this paragraph: {paragraph}"
try_prompt("UNSTRUCTURED PROMPT", p_unstructured)

# 2 Structured prompt with labeled components
p_structured = (
"Instruction: Summarize the following paragraph in one sentence.\n"
"Context: This text is from an academic report on ethical AI practices.\n"
"Constraint: Use formal tone, maximum 40 words.\n"
"Example: Input: 'AI can help doctors.' Output: 'AI supports decision-making when applied responsibly.'\n"
"Output Format: Present your answer as a single bullet point.\n"
f"Paragraph: {paragraph}"
)

try_prompt("STRUCTURED PROMPT", p_structured)
```

```python
p_constrained = (
    "Instruction: Summarize the following paragraph in one sentence.\n"
    "Context: The summary will be used in a university ethics lecture.\n"
    "Constraints: Use academic tone, include the phrase 'responsible AI', limit to 25 words.\n"
    "Output Format: One concise sentence.\n"
    f"Paragraph: {paragraph}"
)
try_prompt("STRUCTURED + CONSTRAINED", p_constrained)
```

```python
p_table = (
    "Instruction: Classify each review as Positive, Negative, or Neutral.\n"
    "Context: Customer feedback from an online shop.\n"
    "Constraints: Output results in a markdown table with two columns (Review, Sentiment).\n"
    "Examples:\n"
    " - 'Loved the product!' Positive\n"
    " - 'Terrible quality.' Negative\n"
    "Output Format: Table.\n"
    "Reviews:\n"
    "1. 'Fast delivery, good quality.'\n"
    "2. 'Not worth the money.'\n"
    "3. 'Okay product, nothing special.'"
)

try_prompt("STRUCTURED PROMPT - TABLE OUTPUT", p_table)
```
