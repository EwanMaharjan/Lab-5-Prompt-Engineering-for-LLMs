```python
import os, random, textwrap
import torch
from transformers import pipeline, AutoModelForCausalLM, AutoTokenizer

# Reproducibility
SEED = 42
random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device selection
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)
```

```python
# choosing a model strategy, picked on of 3 from the lab
```

```python
INSTRUCT_MODEL_ID = "microsoft/Phi-3-mini-4k-instruct"

instruct_tokenizer = AutoTokenizer.from_pretrained(INSTRUCT_MODEL_ID)
instruct_model = AutoModelForCausalLM.from_pretrained(
    INSTRUCT_MODEL_ID,
    torch_dtype="auto",
    device_map="auto" if DEVICE == "cuda" else None
)
instruct_generator = pipeline(
    task="text-generation",
    model=instruct_model,
    tokenizer=instruct_tokenizer,
    return_full_text=False,
)
print("Loaded:", INSTRUCT_MODEL_ID)
```

```python
#helper function (pretty print + safe generation)
```

```python
def show(title, text):
    """Nicely formatted section header + wrapped text."""
    print("\n" + "="*90)
    print(title)
    print("-"*90)
    print(text)
    print("="*90)

def safe_generate(gen_pipeline, prompt, max_new_tokens=80, temperature=0.7, top_p=0.95,
do_sample=True, return_full_text=True):
    """Wrapper around a text-generation pipeline to avoid common pitfalls."""
    try:
        tokenizer = gen_pipeline.tokenizer
        pad_token_id = tokenizer.pad_token_id or tokenizer.eos_token_id
    except Exception:
        pad_token_id = None

    kwargs = dict(
    max_new_tokens=max_new_tokens,
    temperature=temperature,
    top_p=top_p,
    do_sample=do_sample,
    return_full_text=True
    )
    if pad_token_id is not None:
        kwargs["pad_token_id"] = pad_token_id

    output = gen_pipeline(prompt, **kwargs)[0]["generated_text"]
    return output

def try_prompt(label, gen_pipeline, prompt, **gen_kwargs):
    """Run a prompt through a chosen pipeline and pretty-print the result."""
    out = safe_generate(gen_pipeline, prompt, **gen_kwargs)
    show(f"{label} PROMPT", prompt)
    show(f"{label} OUTPUT", out)
```

```python
# EXERCISE 1 - ZERO SHOT, ONE SHOT AND FEW SHOT PROMPTING
```

```python
# Helper function to run prompts with clear labels
def try_prompt(label, prompt):
    out = safe_generate(GEN, prompt, max_new_tokens=120, temperature=0.8)
    show(f"{label} PROMPT", prompt)
    show(f"{label} OUTPUT", out)

# -------------------------------
# Zero-Shot Prompt
# -------------------------------
p_zero = (
"Classify the sentiment of this review in one word (Positive, Negative, or Neutral): "
"'The product broke after one day.'"
)
try_prompt("ZERO-SHOT", p_zero)

# -------------------------------
# One-Shot Prompt
# -------------------------------
p_one = (
"You are a sentiment classifier.\n"
"Example: 'I love this!' Positive\n"
"Now classify in ONE WORD (Positive, Negative, or Neutral): "
"'The product broke after one day.'"
)
try_prompt("ONE-SHOT", p_one)

# -------------------------------
# Few-Shot Prompt
# -------------------------------
p_few = (
"You are a sentiment classifier.\n"
"Examples:\n"
" - 'I love this!' Positive\n"
" - 'This is terrible.' Negative\n"
" - 'It works as expected.' Neutral\n"
"Now classify in ONE WORD (Positive, Negative, or Neutral): "
"'The product broke after one day.'"
)

try_prompt("FEW-SHOT", p_few)
```

```python
results = {}
for label, prompt in {
    "Zero-Shot": p_zero,
    "One-Shot": p_one,
    "Few-Shot": p_few
}.items():
    results[label] = safe_generate(GEN, prompt, max_new_tokens=80, temperature=0.7)

for k, v in results.items():
    show(f"{k} OUTPUT", v)
```

```python
#EXERCISE 5 - EVALUATING PROMPT QUALITY AND ITERATIVE REFINEMENT
```

```python
def score_output(relevance, clarity, structure, factuality, creativity):
    """Simple function to total up 5 evaluation scores."""
    total = relevance + clarity + structure + factuality + creativity
    avg = total / 5
    print(f"Scores -> Relevance:{relevance}, Clarity:{clarity}, Structure:{structure}, "
          f"Factuality:{factuality}, Creativity:{creativity}")
    print(f"Total: {total}/25 | Average: {avg:.2f}")
    
    return avg
```

```python
# Reuse your previous prompt (e.g., structured summarization)
prompt = (
"Instruction: Summarize the following paragraph in one bullet point.\n"
"Context: The text is from an academic article about ethical AI.\n"
"Constraint: Use an academic tone, 40 words maximum.\n"
"Output Format: Bullet point.\n"
"Paragraph: AI systems are widely used, but can propagate bias if trained on unbalanced data."
)
output = safe_generate(GEN, prompt, max_new_tokens=60, temperature=0.7)

show("EVALUATION PROMPT", prompt)
show("EVALUATION OUTPUT", output)
```

```python
 # Example scoring (adjust based on observation)
avg_score = score_output(
relevance=5, # directly answered the task
clarity=4, # easy to read
structure=5, # correct bullet format
factuality=5, # accurate info
creativity=3 # clear but not original
)
```

```python
prompt_refined = (
"Instruction: Summarize the following paragraph in one concise bullet point.\n"
"Context: This is part of a lecture on AI fairness.\n"
"Constraints: Must use the phrase 'ethical AI', limit to 30 words.\n"
"Output Format: Start with ' '.\n"
"Paragraph: AI systems are widely used, but can propagate bias if trained on unbalanced data."
)
output_refined = safe_generate(GEN, prompt_refined, max_new_tokens=60, temperature=0.7)

show("REFINED PROMPT", prompt_refined)
show("REFINED OUTPUT", output_refined)

# Rate again
avg_score_refined = score_output(5, 5, 5, 5, 4)
```
